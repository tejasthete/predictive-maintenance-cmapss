```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# Define column names

columns = [
    "unit_id",
    "cycle",
    "setting_1",
    "setting_2",
    "setting_3"
]

sensor_columns = [f"sensor_{i}" for i in range(1, 22)]

columns.extend(sensor_columns)

# Reload training dataset

train_df = pd.read_csv(
    "../data/train_FD001.txt",
    sep=r"\s+",
    header=None,
    names=columns
)

print("Training data loaded successfully!")
print("Shape:", train_df.shape)
print("\nColumns:")
print(train_df.columns.tolist())
```

```python
#number of engines 
print("Number of engine in Training Data",train_df["unit_id"].nunique())
print("Number of engine in Test  Data",test_df["unit_id"].nunique())
```

```python
#number of cycles
cycles_per_engines = train_df.groupby("unit_id")["cycle"].max()

print(cycles_per_engines.head(10))
```

```python
#Minimum and maximum cycles 

print("Minimum training cycles",cycles_per_engines.min())
print("Maximum training cycles",cycles_per_engines.max())
print("Average training cycles",cycles_per_engines.mean())
```

```python
missing_values = train_df.isnull().sum()

print(missing_values)
```

```python
#sensor statistics
sensor_summary = train_df[sensor_columns].describe().T

sensor_summary
```

```python
# sensors with almost no variation
sensor_std = train_df[sensor_columns].std()

print(sensor_std.sort_values())
```

```python
#plot number of cycles for each engine 

plt.figure(figsize=(12,6))

plt.bar(
    cycles_per_engines.index,
    cycles_per_engines.values
)

plt.xlabel("Engine ID")
plt.ylabel("Number of Cycles")
plt.title("number of cycles for each Training engine")
```

```python
#Distribution of engine lifetimes

plt.figure(figsize=(10,5))

plt.hist(
    cycles_per_engines,
    bins = 20,
    color='green'
)

plt.xlabel("Engine Lifetime Cycles")
plt.ylabel("Number of Engines")
plt.title("Distribution of Engine Lifetimes")

plt.show()
```
